```python
import json
```

```python
def encode(mot, jsons):
    cles = []

    with open(jsons, "r", encoding="utf-8") as f:
        dico = json.load(f)

    for i in mot.split():
        if i in dico:
            cle = dico.get(i)
            if cle is not None:
                cles.append(int(cle))
        else:
            cles = dico.len() + 1
            changes = [i, cles]
            dico = dico + changes
            cles = cles


    return cles
```

```python
if __name__ == "__main__":
    with open("dataset_token.txt","r", encoding="utf-8") as f:
        mots = (f)
    idsss = encode (mots, "vocab.json")
    with open("dataset.txt", "w", encoding="utf-8") as f:
        f.write(idsss)
    print('fin')
    
```
